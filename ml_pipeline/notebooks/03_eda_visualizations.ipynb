# TripSync AI — 03: Comprehensive Exploratory Data Analysis (EDA)
**Objective:** Uncover empirical distributions, geographic patterns, correlation dynamics, and duration-fee interactions across 323 premier Indian tourism destinations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

PROCESSED_FILE = Path("../data/processed/places_cleaned.csv")
df = pd.read_csv(PROCESSED_FILE)
print(f"Loaded {len(df)} cleaned attractions across {df['state'].nunique()} Indian States & UTs.")
df.head(3)
```

## 1. Target Variable (Rating) Statistical Properties

```python
print("Rating Descriptive Statistics:")
print(df['rating'].describe())

plt.figure(figsize=(8, 4))
sns.histplot(df['rating'], kde=True, bins=15, color='#2b5c8f')
plt.title("Distribution of Google Review Ratings")
plt.xlabel("Rating")
plt.show()
```

## 2. Macro-Thematic Experience Categories

```python
category_counts = df['macro_category'].value_counts()
print("Macro Category Distribution:")
print(category_counts)

plt.figure(figsize=(9, 4))
category_counts.plot(kind='bar', color='#2a9d8f', edgecolor='black')
plt.title("Macro Thematic Distribution")
plt.ylabel("Attraction Count")
plt.xticks(rotation=30, ha='right')
plt.show()
```

## 3. Financial & Budget Analysis (INR Fees)

```python
print(f"Free Attractions: {(df['entrance_fee_inr'] == 0).sum()} ({(df['entrance_fee_inr'] == 0).mean()*100:.1f}%)")
print("Budget Tier Breakdown:")
print(df['budget_tier'].value_counts(normalize=True) * 100)
```

## 4. Feature Correlation Matrix

```python
corr_cols = ['rating', 'time_needed_hrs', 'log_entrance_fee', 'log_reviews', 'budget_tier_code', 'has_airport_50km']
plt.figure(figsize=(7, 5))
sns.heatmap(df[corr_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Feature Correlation Matrix")
plt.show()
```
