# TripSync AI — 01: Dataset Acquisition & Quality Inspection
**Objective:** Verify data integrity, inspect dimensional properties, identify null values and duplicate records across the Tourism Destination & Attraction Benchmark Dataset.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/raw")

# Load datasets
places_df = pd.read_csv(DATA_DIR / "tourism_with_id.csv")
ratings_df = pd.read_csv(DATA_DIR / "tourism_rating.csv")
users_df = pd.read_csv(DATA_DIR / "user.csv")

print(f"Places shape:  {places_df.shape}")
print(f"Ratings shape: {ratings_df.shape}")
print(f"Users shape:   {users_df.shape}")
```

```python
# Null value assessment
places_nulls = places_df.isnull().sum()
places_nulls[places_nulls > 0]
```

## 2. Interaction Ratings Inspection (`tourism_rating.csv`)

```python
ratings_df.head()
```

```python
# Duplicate check on user-item interaction pairs
dup_interactions = ratings_df.duplicated(subset=['User_Id', 'Place_Id']).sum()
print(f"Duplicate (User_Id, Place_Id) interactions: {dup_interactions}")
ratings_df['Place_Ratings'].value_counts().sort_index()
```

## 4. Referential Integrity Verification

```python
orphan_places = set(ratings_df['Place_Id']) - set(places_df['Place_Id'])
orphan_users = set(ratings_df['User_Id']) - set(users_df['User_Id'])
print(f"Orphan place interactions: {len(orphan_places)}")
print(f"Orphan user interactions:  {len(orphan_users)}")
assert len(orphan_places) == 0, "Foreign key violation: places"
assert len(orphan_users) == 0, "Foreign key violation: users"
print("Referential integrity perfectly validated!")
```
