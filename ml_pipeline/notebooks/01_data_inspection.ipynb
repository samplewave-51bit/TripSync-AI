# TripSync AI — 01: Indian Tourism Dataset Acquisition & Quality Inspection
**Objective:** Verify data integrity, inspect dimensional properties, assess pan-India geographic coverage, and evaluate attraction types and logistics across the Top Indian Places to Visit Dataset.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/raw")
df = pd.read_csv(DATA_DIR / "top_indian_places.csv")

print(f"Indian Tourism Dataset Shape: {df.shape[0]} attractions x {df.shape[1]} features")
print(f"Total States & UTs Covered: {df['state'].nunique()}")
print(f"Total Cities Represented: {df['city'].nunique()}")
df.head(5)
```

## 1. Data Completeness & Schema Check

```python
nulls = df.isnull().sum()
print(f"Total Missing Values Across Dataset: {nulls.sum()}")
assert nulls.sum() == 0, "Unexpected null values detected!"
print("100% data completeness verified across all 325 Indian destinations.")
```

## 2. Zonal and State-Wise Distribution Across India

```python
print("Zonal Distribution:")
print(df['zone'].value_counts())
print("\nTop 10 States with Most Attractions:")
print(df['state'].value_counts().head(10))
```

## 3. Attraction Types & Categorization

```python
print("Top 15 Attraction Categories:")
print(df['type'].value_counts().head(15))
```

## 4. Financial & Planning Logistics (INR Fees & Duration)

```python
print("Budget Tier Breakdown (INR Entry Fee):")
print(df['budget_tier'].value_counts(normalize=True) * 100)

print("\nGeospatial Coordinate Verification (India Bounds):")
print(f"Latitude Range:  [{df['latitude'].min():.4f}, {df['latitude'].max():.4f}] (South to North)")
print(f"Longitude Range: [{df['longitude'].min():.4f}, {df['longitude'].max():.4f}] (West to East)")

num_cols = ['time_needed_hrs', 'rating', 'entrance_fee_inr', 'reviews_lakhs']
df[num_cols].describe().T[['mean', 'std', 'min', '50%', 'max']]
```
