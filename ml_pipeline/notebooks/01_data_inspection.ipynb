# TripSync AI — 01: Global Dataset Acquisition & Quality Inspection
**Objective:** Verify data integrity, inspect dimensional properties, assess worldwide geographic coverage, and evaluate thematic distributions across the Worldwide Travel Cities Dataset.

```python
import pandas as pd
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/raw")
df = pd.read_csv(DATA_DIR / "worldwide_travel_cities.csv")

print(f"Worldwide Dataset Shape: {df.shape[0]} destinations x {df.shape[1]} features")
print(f"Total Countries Covered: {df['country'].nunique()}")
df.head(3)
```

## 1. Null Assessment and Completeness

```python
nulls = df.isnull().sum()
print(f"Total Missing Values Across Dataset: {nulls.sum()}")
assert nulls.sum() == 0, "Unexpected null values detected!"
print("100% data completeness verified across all 560 cities.")
```

## 2. Global Continental and Regional Distribution

```python
print(df['region'].value_counts())
print("\nTop 15 Countries:")
print(df['country'].value_counts().head(15))
```

## 3. Thematic Experience Ratings (0 - 5 scale)

```python
thematic_cols = ['culture', 'adventure', 'nature', 'beaches', 'nightlife', 'cuisine', 'wellness', 'urban', 'seclusion']
df[thematic_cols].describe().T[['mean', 'std', 'min', '50%', 'max']]
```

## 4. Planning Logistics: Budget and Durations

```python
print("Budget Tier Breakdown:")
print(df['budget_level'].value_counts(normalize=True) * 100)

print("\nGeospatial Coordinate Verification:")
print(f"Latitude Range:  [{df['latitude'].min():.4f}, {df['latitude'].max():.4f}]")
print(f"Longitude Range: [{df['longitude'].min():.4f}, {df['longitude'].max():.4f}]")
```
